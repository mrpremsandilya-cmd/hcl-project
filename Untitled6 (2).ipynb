```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Order_ID': range(1, 51),
    'Item': [
        'Pizza', 'Burger', 'Pasta', 'Sandwich', 'Pizza',
        'Burger', 'Fries', 'Pasta', 'Pizza', 'Sandwich',
        'Burger', 'Pizza', 'Pasta', 'Fries', 'Burger',
        'Sandwich', 'Pizza', 'Pasta', 'Burger', 'Fries',
        'Pizza', 'Sandwich', 'Burger', 'Pasta', 'Pizza',
        'Fries', 'Burger', 'Sandwich', 'Pizza', 'Pasta',
        'Burger', 'Pizza', 'Fries', 'Sandwich', 'Pasta',
        'Pizza', 'Burger', 'Fries', 'Pasta', 'Sandwich',
        'Pizza', 'Burger', 'Pasta', 'Fries', 'Pizza',
        'Sandwich', 'Burger', 'Pasta', 'Pizza', 'Fries'
    ],
    'Quantity': np.random.randint(1, 5, 50),
    'Payment': np.random.choice(
        ['Cash', 'UPI', 'Card'], 50
    ),
    'Date': pd.date_range(
        start='2026-01-01', periods=50, freq='D'
    )
}

data = pd.DataFrame(data)

data.head()
```

```python
prices = {
    'Pizza': 250,
    'Burger': 120,
    'Pasta': 180,
    'Sandwich': 100,
    'Fries': 80
}

categories = {
    'Pizza': 'Main Course',
    'Burger': 'Main Course',
    'Pasta': 'Main Course',
    'Sandwich': 'Snacks',
    'Fries': 'Snacks'
}

data['Price'] = data['Item'].map(prices)

data['Category'] = data['Item'].map(categories)

data['Revenue'] = data['Quantity'] * data['Price']

data.head()
```

```python
print("Dataset Information")
print(data.info())

print("\nDataset Shape")
print(data.shape)

print("\nStatistical Summary")
print(data.describe())

print("\nMissing Values")
print(data.isnull().sum())

print("\nDuplicate Records")
print(data.duplicated().sum())
```

```python
total_orders = data['Order_ID'].nunique()

total_revenue = data['Revenue'].sum()

total_quantity = data['Quantity'].sum()

average_order_value = total_revenue / total_orders

print("Total Orders:", total_orders)

print("Total Revenue: Rs.", total_revenue)

print("Total Items Sold:", total_quantity)

print("Average Order Value: Rs.", round(average_order_value, 2))
```

```python
popular_items = data.groupby('Item')['Quantity'].sum()

print(popular_items.sort_values(ascending=False))
```

```python
category_revenue = data.groupby('Category')['Revenue'].sum()

print(category_revenue)
```

```python
daily_sales = data.groupby('Date')['Revenue'].sum()

print(daily_sales.head())
```

```python
payment_analysis = data['Payment'].value_counts()

payment_analysis.plot(kind='pie', autopct='%1.1f%%',
                      figsize=(7,7))

plt.title('Payment Method Distribution')
plt.ylabel('')
plt.show()
```

```python
best_item = popular_items.idxmax()

print("Best Selling Item:", best_item)
```

```python
data.to_csv('Restaurant_Analysis_Result.csv', index=False)

print("Dataset saved successfully!")
```
